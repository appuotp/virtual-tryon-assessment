```python
import gradio as gr
import json
from PIL import Image

# ---------- Q1 ----------

def load_q1():
    with open("q1_output.json", "r") as f:
        data = json.load(f)

    return json.dumps(data, indent=4)


# ---------- Q2 ----------

def load_q2():
    return (
        "human_parsing_map.png",
        "agnostic_person.png",
        "garment_mask.png"
    )


with gr.Blocks(title="Virtual Try-On Assessment Demo") as demo:

    gr.Markdown(
        """
# 👕 Virtual Try-On Assessment

### Amruth Krishnan M

Master of Computer Applications (MCA)

This demo showcases the completed components of the Virtual Try-On Assessment.
"""
    )

    with gr.Tab("Q1 - Garment Understanding"):

        gr.Markdown("## Garment Attribute Extraction")

        btn1 = gr.Button("Show Q1 Output")

        q1 = gr.Code(language="json")

        btn1.click(
            load_q1,
            outputs=q1
        )


    with gr.Tab("Q2 - Human Parsing"):

        gr.Markdown("## Human Parsing & Segmentation")

        btn2 = gr.Button("Show Q2 Outputs")

        parsing = gr.Image(label="Human Parsing Map")
        agnostic = gr.Image(label="Agnostic Person")
        garment = gr.Image(label="Garment Mask")

        btn2.click(
            load_q2,
            outputs=[parsing, agnostic, garment]
        )


    with gr.Tab("Q3 & Q4"):

        gr.Markdown(
        """
## Q3 - End-to-End Virtual Try-On

- Attempted using CatVTON.
- Repository setup and dependencies were configured.
- Multiple dependency and compatibility issues prevented successful end-to-end inference within the assessment period.

---

## Q4 - Evaluation

Evaluation metrics could not be generated because a final virtual try-on image was not produced.

All implementation attempts and observations have been documented in the repository.
"""
        )


demo.launch(debug=True)
```
